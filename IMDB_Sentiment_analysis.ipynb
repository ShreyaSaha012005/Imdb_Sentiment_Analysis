```python
# ============================================================
#        IMDB Movie Review Sentiment Analysis
#        Complete End-to-End Implementation for Google Colab
# ============================================================

# ─────────────────────────────────────────────────────────────
# STEP 1: INSTALL & IMPORT LIBRARIES
# ─────────────────────────────────────────────────────────────

import io
import re
import string
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
warnings.filterwarnings("ignore")

# NLTK
import nltk
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Scikit-learn
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, confusion_matrix,
    classification_report
)

# TensorFlow / Keras
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Embedding, LSTM, Dense, Dropout, SpatialDropout1D
)
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.callbacks import EarlyStopping

# Google Colab file upload
from google.colab import files

print(" All libraries imported successfully.")
print(f"   TensorFlow version : {tf.__version__}")


# ─────────────────────────────────────────────────────────────
# STEP 2: UPLOAD & LOAD DATASET
# ─────────────────────────────────────────────────────────────

def upload_and_load_dataset():
    """Upload CSV from local system via Colab and return a DataFrame."""
    print("\n Please upload 'IMDB Dataset.csv' from your local system...")
    uploaded = files.upload()                          # Opens Colab file picker

    # Dynamically read whichever file was uploaded
    filename = list(uploaded.keys())[0]
    df = pd.read_csv(io.BytesIO(uploaded[filename]))

    print(f"\n Dataset '{filename}' loaded successfully.")
    print(f"   Shape  : {df.shape}")
    print(f"\n── First 5 rows ──────────────────────────────────────")
    print(df.head())
    print(f"\n── Dataset Info ──────────────────────────────────────")
    print(df.info())
    print(f"\n── Sentiment Distribution ───────────────────────────")
    print(df["sentiment"].value_counts())
    return df


df = upload_and_load_dataset()


# ─────────────────────────────────────────────────────────────
# STEP 3: DATA PREPROCESSING
# ─────────────────────────────────────────────────────────────

# Initialise lemmatizer and stopword set once (global, for speed)
lemmatizer   = WordNetLemmatizer()
STOP_WORDS   = set(stopwords.words("english"))
HTML_PATTERN = re.compile(r"<.*?>")          # matches any HTML tag


def clean_text(text: str) -> str:
    """
    Full preprocessing pipeline applied to a single review string:
      1. Lowercase
      2. Remove HTML tags
      3. Remove punctuation & digits
      4. Tokenise (split on whitespace)
      5. Remove stopwords
      6. Lemmatise each token
    Returns the cleaned string.
    """
    # 1. Lowercase
    text = text.lower()
    # 2. Strip HTML tags  (<br />, <p>, etc.)
    text = HTML_PATTERN.sub(" ", text)
    # 3. Remove punctuation and digits
    text = text.translate(str.maketrans("", "", string.punctuation + string.digits))
    # 4. Tokenise
    tokens = text.split()
    # 5 & 6. Remove stopwords + lemmatise
    tokens = [lemmatizer.lemmatize(t) for t in tokens if t not in STOP_WORDS]
    return " ".join(tokens)


print("\n Cleaning reviews (this may take ~1–2 minutes) ...")
df["clean_review"] = df["review"].apply(clean_text)

# Encode labels  →  positive = 1,  negative = 0
df["label"] = (df["sentiment"] == "positive").astype(int)

print(" Preprocessing complete.")
print("\n── Sample cleaned review ────────────────────────────")
print("ORIGINAL :", df["review"].iloc[0][:200])
print("CLEANED  :", df["clean_review"].iloc[0][:200])


# ─────────────────────────────────────────────────────────────
# STEP 4: TRAIN-TEST SPLIT  (stratified 80 / 20)
# ─────────────────────────────────────────────────────────────

X = df["clean_review"]
y = df["label"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y          # preserves class balance in both splits
)

print(f"\n Train/Test split done.")
print(f"   Training samples : {len(X_train)}")
print(f"   Testing  samples : {len(X_test)}")


# ─────────────────────────────────────────────────────────────
# STEP 5: FEATURE EXTRACTION – TF-IDF  (for ML models)
# ─────────────────────────────────────────────────────────────

print("\n Building TF-IDF matrix ...")
tfidf = TfidfVectorizer(
    max_features=50_000,   # keep top 50 k terms
    ngram_range=(1, 2),    # unigrams + bigrams
    sublinear_tf=True      # apply log(1+tf) scaling
)
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf  = tfidf.transform(X_test)
print(f" TF-IDF matrix shape → train: {X_train_tfidf.shape}, test: {X_test_tfidf.shape}")


# ─────────────────────────────────────────────────────────────
# HELPER: Evaluation utility
# ─────────────────────────────────────────────────────────────

def evaluate_model(name: str, y_true, y_pred, y_prob=None) -> dict:
    """
    Compute and print Accuracy, Precision, Recall, F1, ROC-AUC.
    Returns a dict of scores (for final comparison table).
    """
    acc  = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred)
    rec  = recall_score(y_true, y_pred)
    f1   = f1_score(y_true, y_pred)
    roc  = roc_auc_score(y_true, y_prob) if y_prob is not None else float("nan")

    print(f"\n{'═'*55}")
    print(f"    {name} — Evaluation Report")
    print(f"{'═'*55}")
    print(f"  Accuracy  : {acc:.4f}")
    print(f"  Precision : {prec:.4f}")
    print(f"  Recall    : {rec:.4f}")
    print(f"  F1-Score  : {f1:.4f}")
    print(f"  ROC-AUC   : {roc:.4f}")
    print(f"\n{classification_report(y_true, y_pred, target_names=['Negative','Positive'])}")

    return {"Model": name, "Accuracy": acc, "Precision": prec,
            "Recall": rec, "F1-Score": f1, "ROC-AUC": roc}


def plot_confusion_matrix(name: str, y_true, y_pred):
    """Plot a labelled seaborn heatmap for the confusion matrix."""
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(5, 4))
    sns.heatmap(
        cm, annot=True, fmt="d", cmap="Blues",
        xticklabels=["Negative", "Positive"],
        yticklabels=["Negative", "Positive"]
    )
    plt.title(f"Confusion Matrix — {name}", fontsize=13, fontweight="bold")
    plt.xlabel("Predicted Label")
    plt.ylabel("True Label")
    plt.tight_layout()
    plt.show()


# ─────────────────────────────────────────────────────────────
# STEP 6-A: MODEL — Logistic Regression
# ─────────────────────────────────────────────────────────────

print("\n Training Logistic Regression ...")
lr_model = LogisticRegression(max_iter=1_000, C=5.0, solver="lbfgs")
lr_model.fit(X_train_tfidf, y_train)

lr_pred  = lr_model.predict(X_test_tfidf)
lr_prob  = lr_model.predict_proba(X_test_tfidf)[:, 1]

lr_scores = evaluate_model("Logistic Regression", y_test, lr_pred, lr_prob)
plot_confusion_matrix("Logistic Regression", y_test, lr_pred)


# ─────────────────────────────────────────────────────────────
# STEP 6-B: MODEL — Multinomial Naive Bayes
# ─────────────────────────────────────────────────────────────

print("\n Training Multinomial Naive Bayes ...")
nb_model = MultinomialNB(alpha=0.1)
nb_model.fit(X_train_tfidf, y_train)

nb_pred = nb_model.predict(X_test_tfidf)
nb_prob = nb_model.predict_proba(X_test_tfidf)[:, 1]

nb_scores = evaluate_model("Naive Bayes", y_test, nb_pred, nb_prob)
plot_confusion_matrix("Naive Bayes", y_test, nb_pred)


# ─────────────────────────────────────────────────────────────
# STEP 6-C: MODEL — LSTM (Keras)
# ─────────────────────────────────────────────────────────────

# ── Hyper-parameters ──────────────────────────────────────────
VOCAB_SIZE  = 20_000   # top N words to keep in vocabulary
MAX_LEN     = 200      # pad / truncate each review to this length
EMBED_DIM   = 128      # word-embedding dimension
LSTM_UNITS  = 64       # number of LSTM memory cells
BATCH_SIZE  = 256
EPOCHS      = 10       # EarlyStopping will halt before if val loss stalls

# ── Tokenise text for the LSTM ────────────────────────────────
print("\n Tokenising text for LSTM ...")
tokenizer = Tokenizer(num_words=VOCAB_SIZE, oov_token="<OOV>")
tokenizer.fit_on_texts(X_train)             # fit ONLY on training corpus

# Convert text → integer sequences
train_seqs = tokenizer.texts_to_sequences(X_train)
test_seqs  = tokenizer.texts_to_sequences(X_test)

# Pad / truncate to MAX_LEN (post-padding, post-truncation)
X_train_pad = pad_sequences(train_seqs, maxlen=MAX_LEN,
                             padding="post", truncating="post")
X_test_pad  = pad_sequences(test_seqs,  maxlen=MAX_LEN,
                             padding="post", truncating="post")

y_train_arr = np.array(y_train)
y_test_arr  = np.array(y_test)

print(f" Padded sequences → train: {X_train_pad.shape}, test: {X_test_pad.shape}")

# ── Build LSTM model ──────────────────────────────────────────
def build_lstm_model(vocab_size, embed_dim, max_len, lstm_units):
    """
    Architecture:
      Embedding  →  SpatialDropout1D  →  LSTM  →  Dropout  →  Dense(sigmoid)
    """
    model = Sequential([
        # Embedding layer: maps integer tokens → dense vectors
        Embedding(input_dim=vocab_size, output_dim=embed_dim,
                  input_length=max_len, name="Embedding"),

        # SpatialDropout drops entire feature maps (better for NLP)
        SpatialDropout1D(0.3, name="SpatialDropout"),

        # LSTM: sequence modelling
        LSTM(units=lstm_units, dropout=0.2, recurrent_dropout=0.2,
             name="LSTM"),

        # Standard dropout for regularisation
        Dropout(0.3, name="Dropout"),

        # Binary classification output
        Dense(1, activation="sigmoid", name="Output")
    ])
    model.compile(
        optimizer="adam",
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )
    return model


lstm_model = build_lstm_model(VOCAB_SIZE, EMBED_DIM, MAX_LEN, LSTM_UNITS)
lstm_model.summary()

# ── Early stopping to prevent overfitting ────────────────────
early_stop = EarlyStopping(
    monitor="val_loss",
    patience=2,
    restore_best_weights=True,
    verbose=1
)

# ── Train LSTM ────────────────────────────────────────────────
print("\n Training LSTM model ...")
history = lstm_model.fit(
    X_train_pad, y_train_arr,
    validation_split=0.10,       # 10 % of training set used for validation
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    callbacks=[early_stop],
    verbose=1
)

# ── LSTM Predictions ──────────────────────────────────────────
lstm_prob = lstm_model.predict(X_test_pad, batch_size=BATCH_SIZE, verbose=0).flatten()
lstm_pred = (lstm_prob >= 0.5).astype(int)

lstm_scores = evaluate_model("LSTM", y_test_arr, lstm_pred, lstm_prob)
plot_confusion_matrix("LSTM", y_test_arr, lstm_pred)


# ─────────────────────────────────────────────────────────────
# STEP 7: VISUALISATIONS — LSTM Training Curves
# ─────────────────────────────────────────────────────────────

def plot_lstm_history(history):
    """Plot training vs validation accuracy and loss over epochs."""
    fig, axes = plt.subplots(1, 2, figsize=(13, 4))

    # ── Accuracy ──────────────────────────────────────────────
    axes[0].plot(history.history["accuracy"],     label="Train Accuracy", linewidth=2)
    axes[0].plot(history.history["val_accuracy"], label="Val Accuracy",   linewidth=2, linestyle="--")
    axes[0].set_title("LSTM — Accuracy over Epochs", fontsize=13, fontweight="bold")
    axes[0].set_xlabel("Epoch")
    axes[0].set_ylabel("Accuracy")
    axes[0].legend()
    axes[0].grid(alpha=0.3)

    # ── Loss ──────────────────────────────────────────────────
    axes[1].plot(history.history["loss"],     label="Train Loss", linewidth=2, color="tomato")
    axes[1].plot(history.history["val_loss"], label="Val Loss",   linewidth=2, color="orange", linestyle="--")
    axes[1].set_title("LSTM — Loss over Epochs", fontsize=13, fontweight="bold")
    axes[1].set_xlabel("Epoch")
    axes[1].set_ylabel("Loss")
    axes[1].legend()
    axes[1].grid(alpha=0.3)

    plt.tight_layout()
    plt.show()


plot_lstm_history(history)


# ─────────────────────────────────────────────────────────────
# STEP 8: MODEL COMPARISON TABLE
# ─────────────────────────────────────────────────────────────

all_scores = [lr_scores, nb_scores, lstm_scores]
results_df = pd.DataFrame(all_scores).set_index("Model")

# Round for readability
results_df = results_df.round(4)

print("\n" + "═"*65)
print("    MODEL COMPARISON SUMMARY")
print("═"*65)
print(results_df.to_string())
print("═"*65)

# ── Bar chart comparison ──────────────────────────────────────
results_df.plot(
    kind="bar", figsize=(11, 5),
    colormap="Set2", edgecolor="black", linewidth=0.5
)
plt.title("Model Comparison — All Metrics", fontsize=13, fontweight="bold")
plt.xlabel("Model")
plt.ylabel("Score")
plt.xticks(rotation=0)
plt.ylim(0.80, 1.00)
plt.legend(loc="lower right")
plt.grid(axis="y", alpha=0.4)
plt.tight_layout()
plt.show()


# ─────────────────────────────────────────────────────────────
# STEP 9: FINAL RESULT — Best Performing Model
# ─────────────────────────────────────────────────────────────

best_model = results_df["F1-Score"].idxmax()
best_f1    = results_df.loc[best_model, "F1-Score"]
best_acc   = results_df.loc[best_model, "Accuracy"]
best_roc   = results_df.loc[best_model, "ROC-AUC"]

print("\n" + "★"*55)
print(f"    BEST MODEL  :  {best_model}")
print(f"      F1-Score    :  {best_f1:.4f}")
print(f"      Accuracy    :  {best_acc:.4f}")
print(f"      ROC-AUC     :  {best_roc:.4f}")
print("★"*55)
print("\n Sentiment Analysis pipeline complete!")
```

**IMPROVED LSTM MODEL**

```python
# ============================================================
#        IMPROVED LSTM MODEL (RUN IN NEXT CELL)
# ============================================================

from tensorflow.keras.layers import Bidirectional

#  Updated Hyperparameters
VOCAB_SIZE = 50000
MAX_LEN = 300
EMBED_DIM = 128
BATCH_SIZE = 128
EPOCHS = 15

print("\n Re-tokenising text with improved parameters...")

# Re-tokenize with larger vocab
tokenizer = Tokenizer(num_words=VOCAB_SIZE, oov_token="<OOV>")
tokenizer.fit_on_texts(X_train)

train_seqs = tokenizer.texts_to_sequences(X_train)
test_seqs  = tokenizer.texts_to_sequences(X_test)

# Increase sequence length
X_train_pad = pad_sequences(train_seqs, maxlen=MAX_LEN, padding="post", truncating="post")
X_test_pad  = pad_sequences(test_seqs,  maxlen=MAX_LEN, padding="post", truncating="post")

y_train_arr = np.array(y_train)
y_test_arr  = np.array(y_test)

print(f" New padded shapes → train: {X_train_pad.shape}, test: {X_test_pad.shape}")


# ============================================================
#  IMPROVED MODEL ARCHITECTURE
# ============================================================

def build_improved_lstm():
    model = Sequential([
        Embedding(VOCAB_SIZE, EMBED_DIM, input_length=MAX_LEN),

        SpatialDropout1D(0.3),

        #  Bidirectional LSTM (captures forward + backward context)
        Bidirectional(LSTM(128, return_sequences=True)),

        Dropout(0.3),

        #  Second LSTM layer
        Bidirectional(LSTM(64)),

        Dropout(0.3),

        Dense(64, activation='relu'),

        Dropout(0.3),

        Dense(1, activation='sigmoid')
    ])

    model.compile(
        loss='binary_crossentropy',
        optimizer='adam',
        metrics=['accuracy']
    )

    return model


lstm_model = build_improved_lstm()
lstm_model.summary()


# ============================================================
#  TRAINING
# ============================================================

early_stop = EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True,
    verbose=1
)

print("\n Training IMPROVED LSTM model...")

history = lstm_model.fit(
    X_train_pad,
    y_train_arr,
    validation_split=0.1,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    callbacks=[early_stop],
    verbose=1
)


# ============================================================
#  EVALUATION
# ============================================================

lstm_prob = lstm_model.predict(X_test_pad, batch_size=BATCH_SIZE).flatten()
lstm_pred = (lstm_prob >= 0.5).astype(int)

improved_lstm_scores = evaluate_model("Improved LSTM", y_test_arr, lstm_pred, lstm_prob)
plot_confusion_matrix("Improved LSTM", y_test_arr, lstm_pred)


# ============================================================
#  PLOT TRAINING CURVES
# ============================================================

plot_lstm_history(history)


# ============================================================
#  FINAL RESULT
# ============================================================

print("\n Improved LSTM completed!")
```
